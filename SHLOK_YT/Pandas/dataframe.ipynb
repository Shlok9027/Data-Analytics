```python
# DataFrame Using Dictionary -

import pandas as pd
data = {

'Name': ['Rohit', 'Virat', 'Hardik', 'Abhishek'],

'Age': [37, 35, 30, 22],

'City': ['Mumbai', 'Delhi', 'Ranchi', 'Chennai'],

'Total_Runs': [15000, 12000, 7000, 13000]

}
df = pd.DataFrame(data)

print(df)
```

```python
# DataFrame Using List--

data_list = [

['Rohit', 37, 'Mumbai', 15000],

['Virat', 35, 'Delhi', 12000],

['Hardik', 30, 'Ranchi', 7000],

['Sanju', 22, 'Chennai', 13000]

]
df2 = pd.DataFrame(data_list)

print(df2)
```
