```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv(r'C:\Users\shrey\SHLOK Python Code\ETL\Day-1\employees.csv')

```

```python
df['salary'].fillna(df['salary'].mean(), inplace=True)
```

```python
df['age'].fillna(df['age'].mean(), inplace=True)
```

```python
df
```

```python
df['salary'] = df['salary'].astype(int)
df['age'] = df['age'].astype(int)
```

```python
df
```

```python
def categorical_column(salary):
    if salary > 65000:
        return 'High'
    
    elif salary >= 50000:
        return ' Medium'
    else:
        return 'Low'
    
    
```

```python
df['salary_category'] = df['salary'].apply(categorical_column)
```
